```python
class College: #college class name
    college_name = "Anand college"
    college_address = "jaipur"
    college_sp_no = 44 #class variable
    def display(self):
        print(self.college_name)
        print(self.college_address)
        print(self.college_sp_no)
    def show(self):
        print(self.college_name)
        print(self.college_address)
        print(self.college_sp_no)

student = College()        
```

```python
class College:
    College_name = "Anand" #public
    _College_library = "Anand Library" #protected
    __College_account = 65493297 #private
    def show(self):
        print(self.__College_account)
student = College()
print(student.College_name) 
print(student._College_library)  
student.show()
```
