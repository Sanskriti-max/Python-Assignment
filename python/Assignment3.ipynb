# 1.Print Prime Numbers from a List

```python
nums = [2, 4, 5, 7, 8, 11, 15, 17]

print("Prime numbers are:")

for num in nums:
    if num > 1:
        is_prime = True

        for i in range(2, num):
            if num % i == 0:
                is_prime = False
                break

        if is_prime:
            print(num)
```

# 3. Function to Count Even Numbers and Sum Odd Numbers

```python
def analyze_list(numbers):
    even_count = 0
    odd_sum = 0

    for num in numbers:
        if num % 2 == 0:
            even_count += 1
        else:
            odd_sum += num

    return even_count, odd_sum


nums = [1, 2, 3, 4, 5, 6]

count, total = analyze_list(nums)

print("Count of Even Numbers:", count)
print("Sum of Odd Numbers:", total)
```

# 4. Simple Interest Using Default Arguments

```python
def simple_interest(principal, rate=5, time=1):
    si = (principal * rate * time) / 100
    return si


# Positional Arguments
print("SI =", simple_interest(10000))

# Keyword Arguments
print("SI =", simple_interest(principal=10000, rate=8, time=2))
```

# 5 & 6. Student Class and Grade Calculation

```python
class Student:
    def __init__(self, name, marks):
        self.name = name
        self.marks = marks

    def calculate_grade(self):
        if self.marks >= 90:
            return "A"
        elif self.marks >= 75:
            return "B"
        elif self.marks >= 60:
            return "C"
        else:
            return "D"


s1 = Student("Rahul", 85)
s2 = Student("Priya", 92)

print(s1.name, "Grade:", s1.calculate_grade())
print(s2.name, "Grade:", s2.calculate_grade())
```

# 7. BankAccount Class

```python
class BankAccount:
    def __init__(self, balance):
        self.__balance = balance

    def deposit(self, amount):
        self.__balance += amount
        print("Deposited:", amount)

    def withdraw(self, amount):
        if amount <= self.__balance:
            self.__balance -= amount
            print("Withdrawn:", amount)
        else:
            print("Insufficient Balance")

    def show_balance(self):
        print("Current Balance:", self.__balance)


acc = BankAccount(5000)

acc.deposit(1000)
acc.withdraw(2000)
acc.show_balance()
```

# 8. Exception Handling for Division

```python
try:
    num1 = float(input("Enter first number: "))
    num2 = float(input("Enter second number: "))

    result = num1 / num2
    print("Result =", result)

except ZeroDivisionError:
    print("Error: Division by zero is not allowed.")

except ValueError:
    print("Error: Invalid input. Please enter numbers only.")
```

# 9. File Handling – Student Data

# 10. Read Numbers from File and Calculate Total & Average

```python
try:
    file = open("numbers.txt", "r")

    numbers = []

    for line in file:
        numbers.append(int(line.strip()))

    total = sum(numbers)
    average = total / len(numbers)

    print("Total =", total)
    print("Average =", average)

    file.close()

except FileNotFoundError:
    print("Error: File does not exist.")
```
