```python
def factorial(n):
    if n == 0 or n == 1:
        return 1 
    return n*factorial(n-1)
print(factorial(5))
```

```python
def apply_twice(func, value):
    return func(func(value))
def double(x):
    return x*2
print(apply_twice(double, 3))    
```
